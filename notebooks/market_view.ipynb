# Market view — the latest days, every timeframe's structure, and the news

One instrument on a **1-minute chart**, ending at its **latest closed candle**: trendlines, support /
resistance zones and fair value gaps from **1m, 5m, 15m, 1h and 4h**, and the economic calendar along
the bottom. No strategy and no trades.

Everything you might change is in **`market_view_config.py`** — this notebook states nothing.

| want | set in `market_view_config.py` |
|---|---|
| another instrument (BTC, NASDAQ, GOLD, USDJPY, ...) | `INSTRUMENT = "GOLD"` |
| an instrument that is not listed | add it to `INSTRUMENTS` as `dict(symbol=..., source=...)` |
| more or fewer days | `DAYS` |
| other timeframes for lines / gaps / zones | `LINE_TFS`, `FVG_TFS`, `SR_TFS` |
| what is switched on when the chart opens | `OPEN_WITH` (the rest is one click away in the legend) |
| which news, and from which level | `NEWS`, `NEWS_SHOW` |

The first run for an instrument downloads what its cache is missing (a minute or two); after that a
run takes about 20 seconds.

```python
import sys, os, warnings; warnings.filterwarnings('ignore')
for _p in (os.path.abspath('..'), os.getcwd()):
    if os.path.isfile(os.path.join(_p, 'quant', '__init__.py')) and _p not in sys.path:
        sys.path.insert(0, _p)

from quant.research import market_view
from quant.research.workspace import market_view_config
from quant.viz import AS_OF_JS, open_in_browser

cfg = market_view_config()
print(f"instrument  {cfg.INSTRUMENT}      (available: {', '.join(cfg.INSTRUMENTS)})")
```

## 1 · The data and the news

```python
view = market_view(cfg)
view.summary()
```

## 2 · The chart

Opens in a new browser tab. Bottom left: **Rewind** — arm it, click a candle, and every line, gap
and zone snaps back to what was knowable at that candle. Bottom right: the **News** level. Top: the
`S/R:` and `FVGs:` rows show one timeframe at a time.

```python
fig = view.figure()
open_in_browser(fig, f"market_{cfg.INSTRUMENT}.html", extra_js=AS_OF_JS if cfg.REWIND else None)
```
