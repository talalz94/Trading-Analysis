# Evaluate — is it real?

Runs the strategy named in `evaluate_config.py`, and every arm of its `COMPARE`, over each quarter
of `TRAIN` and `FORWARD` — about a minute for 19 quarters. Choose between arms on **train**; the
verdict is **forward**, which the choice never saw.

This is the check to run before believing the lab. Notebook 21's strategy made +4R on its 20 days
and loses 0.16R a trade over four and a half years.

```python
import sys, os, warnings; warnings.filterwarnings('ignore')
for _p in (os.path.abspath('..'), os.getcwd()):
    if os.path.isfile(os.path.join(_p, 'quant', '__init__.py')) and _p not in sys.path:
        sys.path.insert(0, _p)

from quant.research.workspace import evaluate_config
from experiments.evaluate import evaluate

cfg = evaluate_config()
print(f"strategy  {cfg.STRATEGY}   {cfg.SYMBOL} {cfg.TF}")
print(f"train {cfg.TRAIN[0]}..{cfg.TRAIN[1]}   forward {cfg.FORWARD[0]}..{cfg.FORWARD[1]}")
print(f"arms      {', '.join(cfg.COMPARE) or '(none)'}")
```

## 1 · Run it

```python
ev = evaluate(cfg)
```

## 2 · The strategy itself

`R / trade` after costs is the number that matters. The zero-cost row says whether a loss is the
rule or the costs: a strategy positive at zero cost and negative after it has an entry that is
not worth what it costs to take — a different problem from an entry with nothing in it.

```python
ev.baseline()
```

```python
ev.by_quarter()
```

## 3 · Its questions

Each arm against the strategy as set, per quarter. Read in this order:

1. **the verdict line** — if the train ranking does not survive forward, or no `|t|` reaches
   ~2.5, the table is noise however large a number in it looks;
2. **forward dR and forward beat** — the arm's effect on quarters the choice never saw;
3. **fwd trim3** — the same with each run's three best trades removed; a big gap from forward dR
   means one or two outsized trades are doing the work.
